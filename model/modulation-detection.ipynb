```python
%%writefile train_models.py

# ============================================================
# RadioML 2018.01A
# 4-Family Modulation Classification
#
# Classes:
#   FSK  -> GMSK
#   QAM  -> 16/32/64/128/256QAM
#   QPSK -> QPSK
#   PSK  -> BPSK/8PSK/16PSK/32PSK
#
# Models:
#   1. Baseline Additive Attention
#   2. Causal Self-Attention
#   3. Residual CNN + Bidirectional Multi-Head Attention
#
# Kaggle-compatible: NO argparse
# ============================================================

import os
import json
import random
import time

import h5py
import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader

from tqdm.auto import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix,
)


# ============================================================
# CONFIGURATION
# ============================================================

HDF5_PATH = (
    "/kaggle/input/datasets/pinxau1000/"
    "radioml2018/GOLD_XYZ_OSC.0001_1024.hdf5"
)

OUTPUT_DIR = "/kaggle/working/radioml_results"

INPUT_LENGTH = 1024
INPUT_CHANNELS = 2
NUM_CLASSES = 4

CLASS_NAMES = [
    "FSK",
    "QAM",
    "QPSK",
    "PSK",
]

CLASS_TO_INDEX = {
    "FSK": 0,
    "QAM": 1,
    "QPSK": 2,
    "PSK": 3,
}

INDEX_TO_CLASS = {
    0: "FSK",
    1: "QAM",
    2: "QPSK",
    3: "PSK",
}


# ============================================================
# RADIO ML 2018 FIXED CLASS ORDER
# ============================================================

RADIO_ML_CLASSES = [
    "OOK",
    "4ASK",
    "8ASK",
    "BPSK",
    "QPSK",
    "8PSK",
    "16PSK",
    "32PSK",
    "16APSK",
    "32APSK",
    "64APSK",
    "128APSK",
    "16QAM",
    "32QAM",
    "64QAM",
    "128QAM",
    "256QAM",
    "AM-SSB-WC",
    "AM-SSB-SC",
    "AM-DSB-WC",
    "AM-DSB-SC",
    "FM",
    "GMSK",
    "OQPSK",
]


# ============================================================
# FOUR FAMILY MAPPING
# ============================================================

FAMILY_MAPPING = {
    "FSK": [
        "GMSK"
    ],

    "QAM": [
        "16QAM",
        "32QAM",
        "64QAM",
        "128QAM",
        "256QAM"
    ],

    "QPSK": [
        "QPSK"
    ],

    "PSK": [
        "BPSK",
        "8PSK",
        "16PSK",
        "32PSK"
    ],
}


# ============================================================
# REPRODUCIBILITY
# ============================================================

def set_seed(seed=42):

    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================================
# DEVICE
# ============================================================

def get_device():

    if torch.cuda.is_available():

        device = torch.device("cuda")

        print(
            "GPU:",
            torch.cuda.get_device_name(0)
        )

    else:

        device = torch.device("cpu")

        print("Using CPU")

    return device


# ============================================================
# DATASET INSPECTION
# ============================================================

def inspect_dataset():

    print("\n" + "=" * 70)
    print("RADIO ML 2018.01A DATASET")
    print("=" * 70)

    if not os.path.exists(HDF5_PATH):

        raise FileNotFoundError(
            f"\nDataset not found:\n{HDF5_PATH}"
        )

    with h5py.File(HDF5_PATH, "r") as h5:

        for key in h5.keys():

            print(
                f"{key}: "
                f"shape={h5[key].shape}, "
                f"dtype={h5[key].dtype}"
            )

    print("=" * 70)


# ============================================================
# PREPARE DATA
# ============================================================

def prepare_four_class_indices(
    max_samples=12000,
    seed=42,
):

    print("\nReading Y and Z...")

    with h5py.File(HDF5_PATH, "r") as h5:

        Y = h5["Y"][:]
        Z = h5["Z"][:].reshape(-1)

    total_samples = Y.shape[0]

    print(
        "Total samples:",
        total_samples
    )

    original_labels = np.argmax(
        Y,
        axis=1
    )

    original_class_to_index = {
        name: idx
        for idx, name in enumerate(RADIO_ML_CLASSES)
    }

    class_to_family = {}

    for family, modulation_list in FAMILY_MAPPING.items():

        for modulation in modulation_list:

            original_index = (
                original_class_to_index[modulation]
            )

            class_to_family[
                original_index
            ] = CLASS_TO_INDEX[family]

    selected_indices = []

    for original_index in class_to_family:

        indices = np.where(
            original_labels == original_index
        )[0]

        selected_indices.extend(
            indices.tolist()
        )

    selected_indices = np.asarray(
        selected_indices,
        dtype=np.int64
    )

    rng = np.random.default_rng(seed)

    rng.shuffle(selected_indices)

    if max_samples is not None:

        selected_indices = (
            selected_indices[
                :min(
                    max_samples,
                    len(selected_indices)
                )
            ]
        )

    selected_original_labels = (
        original_labels[selected_indices]
    )

    family_labels = np.asarray(
        [
            class_to_family[x]
            for x in selected_original_labels
        ],
        dtype=np.int64
    )

    selected_snr = Z[selected_indices]

    print(
        "\nSelected samples:",
        len(selected_indices)
    )

    print("\nFamily distribution:")

    for i, name in enumerate(CLASS_NAMES):

        count = np.sum(
            family_labels == i
        )

        print(
            f"{name:>5}: {count}"
        )

    print("\nSNR distribution:")

    unique_snr, snr_counts = np.unique(
        selected_snr,
        return_counts=True
    )

    for snr, count in zip(
        unique_snr,
        snr_counts
    ):

        print(
            f"SNR {snr:>5}: {count}"
        )

    return (
        selected_indices,
        family_labels,
        selected_snr,
    )


# ============================================================
# HDF5 DATASET
# ============================================================

class RadioMLDataset(Dataset):

    def __init__(
        self,
        indices,
        labels,
    ):

        self.indices = np.asarray(
            indices,
            dtype=np.int64
        )

        self.labels = np.asarray(
            labels,
            dtype=np.int64
        )

        self.h5_file = None

    def open_file(self):

        if self.h5_file is None:

            self.h5_file = h5py.File(
                HDF5_PATH,
                "r"
            )

    def __len__(self):

        return len(self.indices)

    def __getitem__(self, idx):

        self.open_file()

        real_index = self.indices[idx]

        signal = self.h5_file["X"][
            real_index
        ]

        signal = np.asarray(
            signal,
            dtype=np.float32
        )

        # ----------------------------------------------------
        # Per-channel normalization
        # ----------------------------------------------------

        mean = signal.mean(
            axis=0,
            keepdims=True
        )

        std = signal.std(
            axis=0,
            keepdims=True
        )

        signal = (
            signal - mean
        ) / (
            std + 1e-6
        )

        # (1024, 2) -> (2, 1024)

        signal = signal.T

        signal = torch.from_numpy(
            signal
        ).float()

        label = torch.tensor(
            self.labels[idx],
            dtype=torch.long
        )

        return signal, label


# ============================================================
# RESIDUAL BLOCK
# ============================================================

class ResidualBlock(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels,
        kernel_size=7,
        dropout=0.10,
    ):

        super().__init__()

        padding = kernel_size // 2

        self.conv1 = nn.Conv1d(
            in_channels,
            out_channels,
            kernel_size,
            padding=padding,
            bias=False,
        )

        self.bn1 = nn.BatchNorm1d(
            out_channels
        )

        self.conv2 = nn.Conv1d(
            out_channels,
            out_channels,
            kernel_size,
            padding=padding,
            bias=False,
        )

        self.bn2 = nn.BatchNorm1d(
            out_channels
        )

        self.dropout = nn.Dropout(
            dropout
        )

        if in_channels != out_channels:

            self.shortcut = nn.Sequential(

                nn.Conv1d(
                    in_channels,
                    out_channels,
                    kernel_size=1,
                    bias=False,
                ),

                nn.BatchNorm1d(
                    out_channels
                )
            )

        else:

            self.shortcut = nn.Identity()

    def forward(self, x):

        identity = self.shortcut(x)

        out = self.conv1(x)
        out = self.bn1(out)
        out = F.gelu(out)

        out = self.dropout(out)

        out = self.conv2(out)
        out = self.bn2(out)

        out = out + identity

        out = F.gelu(out)

        return out


# ============================================================
# CNN FEATURE EXTRACTOR
# ============================================================

class SignalCNN(nn.Module):

    def __init__(self):

        super().__init__()

        self.stem = nn.Sequential(

            nn.Conv1d(
                2,
                64,
                kernel_size=7,
                padding=3,
                bias=False
            ),

            nn.BatchNorm1d(64),

            nn.GELU(),

            nn.MaxPool1d(2)
        )

        self.block1 = ResidualBlock(
            64,
            128,
            kernel_size=7
        )

        self.pool1 = nn.MaxPool1d(2)

        self.block2 = ResidualBlock(
            128,
            256,
            kernel_size=5
        )

        self.pool2 = nn.MaxPool1d(2)

        self.block3 = ResidualBlock(
            256,
            256,
            kernel_size=3
        )

    def forward(self, x):

        x = self.stem(x)

        # 1024 -> 512

        x = self.block1(x)

        x = self.pool1(x)

        # 512 -> 256

        x = self.block2(x)

        x = self.pool2(x)

        # 256 -> 128

        x = self.block3(x)

        return x


# ============================================================
# BASELINE ADDITIVE ATTENTION
# ============================================================

class BaselineAttention(nn.Module):

    def __init__(
        self,
        feature_dim=256,
        attention_dim=128,
    ):

        super().__init__()

        self.score = nn.Sequential(

            nn.Linear(
                feature_dim,
                attention_dim
            ),

            nn.Tanh(),

            nn.Linear(
                attention_dim,
                1
            )
        )

    def forward(self, x):

        # x = B x T x C

        scores = self.score(x)

        weights = torch.softmax(
            scores,
            dim=1
        )

        context = torch.sum(
            x * weights,
            dim=1
        )

        return context, weights


# ============================================================
# CAUSAL SELF ATTENTION
# ============================================================

class CausalAttention(nn.Module):

    def __init__(
        self,
        feature_dim=256,
        heads=8,
        dropout=0.10,
    ):

        super().__init__()

        self.attention = nn.MultiheadAttention(
            embed_dim=feature_dim,
            num_heads=heads,
            dropout=dropout,
            batch_first=True
        )

        self.norm1 = nn.LayerNorm(
            feature_dim
        )

        self.ffn = nn.Sequential(

            nn.Linear(
                feature_dim,
                feature_dim * 4
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                feature_dim * 4,
                feature_dim
            )
        )

        self.norm2 = nn.LayerNorm(
            feature_dim
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(self, x):

        batch_size, seq_len, _ = x.shape

        # ----------------------------------------------------
        # Causal mask
        #
        # Position t cannot attend to future positions.
        # ----------------------------------------------------

        mask = torch.triu(
            torch.ones(
                seq_len,
                seq_len,
                device=x.device,
                dtype=torch.bool
            ),
            diagonal=1
        )

        attended, weights = (
            self.attention(
                x,
                x,
                x,
                attn_mask=mask,
                need_weights=True
            )
        )

        x = self.norm1(
            x + self.dropout(attended)
        )

        feed_forward = self.ffn(x)

        x = self.norm2(
            x + self.dropout(feed_forward)
        )

        # Global average pooling

        context = x.mean(
            dim=1
        )

        return context, weights


# ============================================================
# BIDIRECTIONAL MULTI-HEAD ATTENTION
# ============================================================

class MultiHeadAttentionBlock(nn.Module):

    def __init__(
        self,
        feature_dim=256,
        heads=8,
        dropout=0.10,
    ):

        super().__init__()

        self.attention = nn.MultiheadAttention(
            embed_dim=feature_dim,
            num_heads=heads,
            dropout=dropout,
            batch_first=True
        )

        self.norm1 = nn.LayerNorm(
            feature_dim
        )

        self.ffn = nn.Sequential(

            nn.Linear(
                feature_dim,
                feature_dim * 4
            ),

            nn.GELU(),

            nn.Dropout(dropout),

            nn.Linear(
                feature_dim * 4,
                feature_dim
            )
        )

        self.norm2 = nn.LayerNorm(
            feature_dim
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(self, x):

        attended, weights = (
            self.attention(
                x,
                x,
                x,
                need_weights=True
            )
        )

        x = self.norm1(
            x + self.dropout(attended)
        )

        feed_forward = self.ffn(x)

        x = self.norm2(
            x + self.dropout(feed_forward)
        )

        context = x.mean(
            dim=1
        )

        return context, weights


# ============================================================
# COMPLETE MODEL
# ============================================================

class ModulationClassifier(nn.Module):

    def __init__(
        self,
        attention_type="mha",
    ):

        super().__init__()

        self.attention_type = attention_type

        self.cnn = SignalCNN()

        # ----------------------------------------------------
        # Positional encoding
        # ----------------------------------------------------

        self.position = nn.Parameter(
            torch.randn(
                1,
                128,
                256
            ) * 0.02
        )

        if attention_type == "baseline":

            self.attention = (
                BaselineAttention(
                    256,
                    128
                )
            )

        elif attention_type == "causal":

            self.attention = (
                CausalAttention(
                    256,
                    heads=8
                )
            )

        elif attention_type == "mha":

            self.attention = (
                MultiHeadAttentionBlock(
                    256,
                    heads=8
                )
            )

        else:

            raise ValueError(
                f"Unknown attention type: "
                f"{attention_type}"
            )

        self.classifier = nn.Sequential(

            nn.Linear(
                256,
                256
            ),

            nn.LayerNorm(256),

            nn.GELU(),

            nn.Dropout(0.35),

            nn.Linear(
                256,
                NUM_CLASSES
            )
        )

    def forward(
        self,
        x,
        return_attention=False,
    ):

        x = self.cnn(x)

        # B x 256 x 128
        #
        # -> B x 128 x 256

        x = x.transpose(
            1,
            2
        )

        x = x + self.position

        context, attention_weights = (
            self.attention(x)
        )

        logits = self.classifier(
            context
        )

        if return_attention:

            return (
                logits,
                attention_weights
            )

        return logits


# ============================================================
# CLASS WEIGHTS
# ============================================================

def get_class_weights(labels):

    counts = np.bincount(
        labels,
        minlength=NUM_CLASSES
    )

    counts = np.maximum(
        counts,
        1
    )

    total = len(labels)

    weights = total / (
        NUM_CLASSES * counts
    )

    return torch.tensor(
        weights,
        dtype=torch.float32
    )


# ============================================================
# TRAIN ONE EPOCH
# ============================================================

def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device,
):

    model.train()

    total_loss = 0.0

    predictions = []
    targets = []

    for signals, labels in tqdm(
        loader,
        desc="Train",
        leave=False
    ):

        signals = signals.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        outputs = model(
            signals
        )

        loss = criterion(
            outputs,
            labels
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        total_loss += (
            loss.item()
            * signals.size(0)
        )

        pred = torch.argmax(
            outputs,
            dim=1
        )

        predictions.extend(
            pred.detach()
            .cpu()
            .numpy()
        )

        targets.extend(
            labels.detach()
            .cpu()
            .numpy()
        )

    loss = (
        total_loss
        / len(loader.dataset)
    )

    accuracy = accuracy_score(
        targets,
        predictions
    )

    return loss, accuracy


# ============================================================
# EVALUATION
# ============================================================

@torch.no_grad()
def evaluate(
    model,
    loader,
    criterion,
    device,
):

    model.eval()

    total_loss = 0.0

    predictions = []
    targets = []

    for signals, labels in tqdm(
        loader,
        desc="Evaluate",
        leave=False
    ):

        signals = signals.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        outputs = model(
            signals
        )

        loss = criterion(
            outputs,
            labels
        )

        total_loss += (
            loss.item()
            * signals.size(0)
        )

        pred = torch.argmax(
            outputs,
            dim=1
        )

        predictions.extend(
            pred.cpu().numpy()
        )

        targets.extend(
            labels.cpu().numpy()
        )

    loss = (
        total_loss
        / len(loader.dataset)
    )

    accuracy = accuracy_score(
        targets,
        predictions
    )

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            targets,
            predictions,
            average="weighted",
            zero_division=0
        )
    )

    return (
        loss,
        accuracy,
        precision,
        recall,
        f1,
        np.asarray(targets),
        np.asarray(predictions),
    )


# ============================================================
# TRAIN MODEL
# ============================================================

def train_model(
    model_name,
    train_loader,
    val_loader,
    test_loader,
    train_labels,
    device,
    epochs=15,
    learning_rate=3e-4,
):

    print("\n")
    print("=" * 70)
    print(
        f"TRAINING MODEL: {model_name.upper()}"
    )
    print("=" * 70)

    model = ModulationClassifier(
        attention_type=model_name
    ).to(device)

    # --------------------------------------------------------
    # Label smoothing + class weighting
    # --------------------------------------------------------

    class_weights = get_class_weights(
        train_labels
    ).to(device)

    criterion = nn.CrossEntropyLoss(
        weight=class_weights,
        label_smoothing=0.05
    )

    # --------------------------------------------------------
    # AdamW
    # --------------------------------------------------------

    optimizer = optim.AdamW(
        model.parameters(),
        lr=learning_rate,
        weight_decay=1e-4
    )

    # --------------------------------------------------------
    # Cosine scheduler
    # --------------------------------------------------------

    scheduler = optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=epochs,
        eta_min=1e-6
    )

    best_accuracy = 0.0
    best_epoch = 0

    model_dir = os.path.join(
        OUTPUT_DIR,
        "models"
    )

    os.makedirs(
        model_dir,
        exist_ok=True
    )

    checkpoint_path = os.path.join(
        model_dir,
        f"{model_name}_best.pt"
    )

    history = []

    for epoch in range(
        1,
        epochs + 1
    ):

        start = time.time()

        print(
            f"\nEpoch {epoch}/{epochs}"
        )

        train_loss, train_acc = (
            train_one_epoch(
                model,
                train_loader,
                criterion,
                optimizer,
                device
            )
        )

        (
            val_loss,
            val_acc,
            val_precision,
            val_recall,
            val_f1,
            _,
            _,
        ) = evaluate(
            model,
            val_loader,
            criterion,
            device
        )

        scheduler.step()

        lr = optimizer.param_groups[0]["lr"]

        elapsed = time.time() - start

        print(
            f"Train Loss : {train_loss:.4f}"
        )

        print(
            f"Train Acc  : {train_acc:.4f}"
        )

        print(
            f"Val Loss   : {val_loss:.4f}"
        )

        print(
            f"Val Acc    : {val_acc:.4f}"
        )

        print(
            f"Val F1     : {val_f1:.4f}"
        )

        print(
            f"LR         : {lr:.7f}"
        )

        print(
            f"Time       : {elapsed:.1f}s"
        )

        row = {
            "epoch": epoch,
            "train_loss": float(train_loss),
            "train_accuracy": float(train_acc),
            "val_loss": float(val_loss),
            "val_accuracy": float(val_acc),
            "val_precision": float(val_precision),
            "val_recall": float(val_recall),
            "val_f1": float(val_f1),
            "learning_rate": float(lr),
        }

        history.append(row)

        if val_acc > best_accuracy:

            best_accuracy = val_acc
            best_epoch = epoch

            torch.save(
                {
                    "model_state_dict":
                        model.state_dict(),

                    "attention_type":
                        model_name,

                    "best_val_accuracy":
                        float(val_acc),

                    "epoch":
                        epoch,

                    "class_names":
                        CLASS_NAMES,
                },
                checkpoint_path
            )

            print(
                ">>> Best model saved"
            )

    # --------------------------------------------------------
    # Load best model
    # --------------------------------------------------------

    checkpoint = torch.load(
        checkpoint_path,
        map_location=device
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    # --------------------------------------------------------
    # Final test
    # --------------------------------------------------------

    (
        test_loss,
        test_accuracy,
        test_precision,
        test_recall,
        test_f1,
        y_true,
        y_pred,
    ) = evaluate(
        model,
        test_loader,
        criterion,
        device
    )

    print("\n")
    print("=" * 70)
    print(
        f"FINAL RESULTS: {model_name.upper()}"
    )
    print("=" * 70)

    print(
        f"Best Epoch : {best_epoch}"
    )

    print(
        f"Best Val   : {best_accuracy:.4f}"
    )

    print(
        f"Test Acc   : {test_accuracy:.4f}"
    )

    print(
        f"Precision   : {test_precision:.4f}"
    )

    print(
        f"Recall      : {test_recall:.4f}"
    )

    print(
        f"F1          : {test_f1:.4f}"
    )

    print("\nClassification Report:")

    report = classification_report(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES)),
        target_names=CLASS_NAMES,
        output_dict=True,
        zero_division=0
    )

    print(
        classification_report(
            y_true,
            y_pred,
            labels=list(range(NUM_CLASSES)),
            target_names=CLASS_NAMES,
            zero_division=0
        )
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES))
    )

    result = {
        "model": model_name,
        "best_epoch": best_epoch,
        "best_val_accuracy":
            float(best_accuracy),
        "test_loss":
            float(test_loss),
        "test_accuracy":
            float(test_accuracy),
        "test_precision":
            float(test_precision),
        "test_recall":
            float(test_recall),
        "test_f1":
            float(test_f1),
        "classification_report":
            report,
        "confusion_matrix":
            cm.tolist(),
        "history":
            history,
    }

    return result


# ============================================================
# MAIN
# ============================================================

def main():

    # ========================================================
    # EXPERIMENT CONFIGURATION
    # ========================================================

    SEED = 42

    # --------------------------------------------------------
    # SANITY MODE
    #
    # Change to None for larger/full experiment.
    # --------------------------------------------------------

    MAX_SAMPLES = 12000

    EPOCHS = 15

    BATCH_SIZE = 128

    LEARNING_RATE = 3e-4

    NUM_WORKERS = 2

    # --------------------------------------------------------
    # Choose model here for individual training.
    #
    # baseline
    # causal
    # mha
    # --------------------------------------------------------

    MODEL_TO_TRAIN = "mha"

    # ========================================================
    # SETUP
    # ========================================================

    set_seed(SEED)

    os.makedirs(
        OUTPUT_DIR,
        exist_ok=True
    )

    device = get_device()

    inspect_dataset()

    # ========================================================
    # PREPARE DATA
    # ========================================================

    (
        selected_indices,
        family_labels,
        selected_snr,
    ) = prepare_four_class_indices(
        max_samples=MAX_SAMPLES,
        seed=SEED
    )

    # ========================================================
    # TRAIN / VAL / TEST
    #
    # 70 / 15 / 15
    # ========================================================

    (
        train_indices,
        temp_indices,
        train_labels,
        temp_labels,
        train_snr,
        temp_snr,
    ) = train_test_split(
        selected_indices,
        family_labels,
        selected_snr,
        test_size=0.30,
        random_state=SEED,
        stratify=family_labels
    )

    (
        val_indices,
        test_indices,
        val_labels,
        test_labels,
        val_snr,
        test_snr,
    ) = train_test_split(
        temp_indices,
        temp_labels,
        temp_snr,
        test_size=0.50,
        random_state=SEED,
        stratify=temp_labels
    )

    print("\n")
    print("=" * 70)
    print("DATASET SPLIT")
    print("=" * 70)

    print(
        "Train      :",
        len(train_indices)
    )

    print(
        "Validation :",
        len(val_indices)
    )

    print(
        "Test       :",
        len(test_indices)
    )

    # ========================================================
    # DATASETS
    # ========================================================

    train_dataset = RadioMLDataset(
        train_indices,
        train_labels
    )

    val_dataset = RadioMLDataset(
        val_indices,
        val_labels
    )

    test_dataset = RadioMLDataset(
        test_indices,
        test_labels
    )

    # ========================================================
    # DATALOADERS
    # ========================================================

    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available()
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available()
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available()
    )

    # ========================================================
    # TRAIN
    # ========================================================

    result = train_model(
        MODEL_TO_TRAIN,
        train_loader,
        val_loader,
        test_loader,
        train_labels,
        device,
        epochs=EPOCHS,
        learning_rate=LEARNING_RATE
    )

    # ========================================================
    # SAVE RESULT
    # ========================================================

    result_path = os.path.join(
        OUTPUT_DIR,
        f"{MODEL_TO_TRAIN}_results.json"
    )

    with open(
        result_path,
        "w"
    ) as f:

        json.dump(
            result,
            f,
            indent=4
        )

    # --------------------------------------------------------
    # Save history CSV
    # --------------------------------------------------------

    history_df = pd.DataFrame(
        result["history"]
    )

    history_path = os.path.join(
        OUTPUT_DIR,
        f"{MODEL_TO_TRAIN}_history.csv"
    )

    history_df.to_csv(
        history_path,
        index=False
    )

    print("\n")
    print("=" * 70)
    print("TRAINING COMPLETE")
    print("=" * 70)

    print(
        "Result:",
        result_path
    )

    print(
        "History:",
        history_path
    )


if __name__ == "__main__":
    main()
```

```python
%%writefile visualize.py

# ============================================================
# RadioML 2018.01A
# VISUALIZATION
# ============================================================

import os
import json

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import ConfusionMatrixDisplay


# ============================================================
# CONFIGURATION
# ============================================================

RESULTS_DIR = "/kaggle/working/radioml_results"

CLASS_NAMES = [
    "FSK",
    "QAM",
    "QPSK",
    "PSK",
]


# ============================================================
# LOAD COMPARISON
# ============================================================

def load_comparison():

    path = os.path.join(
        RESULTS_DIR,
        "model_comparison.csv"
    )

    if not os.path.exists(path):

        print(
            "model_comparison.csv not found."
        )

        print(
            "Run experiment.py first."
        )

        return None

    df = pd.read_csv(
        path
    )

    return df


# ============================================================
# LOAD RESULT JSON
# ============================================================

def load_result(model_name):

    path = os.path.join(
        RESULTS_DIR,
        f"{model_name}_results.json"
    )

    if not os.path.exists(path):

        print(
            f"Missing result file: {path}"
        )

        return None

    with open(
        path,
        "r"
    ) as f:

        return json.load(f)


# ============================================================
# MODEL ACCURACY COMPARISON
# ============================================================

def plot_accuracy_comparison(df):

    plt.figure(
        figsize=(9, 6)
    )

    plt.bar(
        df["Model"],
        df["Test Accuracy"]
    )

    plt.ylabel(
        "Test Accuracy"
    )

    plt.xlabel(
        "Model"
    )

    plt.title(
        "RadioML 2018.01A - Model Accuracy Comparison"
    )

    plt.ylim(
        0,
        1
    )

    for i, value in enumerate(
        df["Test Accuracy"]
    ):

        plt.text(
            i,
            value + 0.02,
            f"{value:.3f}",
            ha="center"
        )

    plt.tight_layout()

    path = os.path.join(
        RESULTS_DIR,
        "accuracy_comparison.png"
    )

    plt.savefig(
        path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()

    print(
        "Saved:",
        path
    )


# ============================================================
# F1 COMPARISON
# ============================================================

def plot_f1_comparison(df):

    plt.figure(
        figsize=(9, 6)
    )

    plt.bar(
        df["Model"],
        df["F1"]
    )

    plt.ylabel(
        "Weighted F1 Score"
    )

    plt.xlabel(
        "Model"
    )

    plt.title(
        "Weighted F1 Comparison"
    )

    plt.ylim(
        0,
        1
    )

    for i, value in enumerate(
        df["F1"]
    ):

        plt.text(
            i,
            value + 0.02,
            f"{value:.3f}",
            ha="center"
        )

    plt.tight_layout()

    path = os.path.join(
        RESULTS_DIR,
        "f1_comparison.png"
    )

    plt.savefig(
        path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()

    print(
        "Saved:",
        path
    )


# ============================================================
# TRAINING CURVES
# ============================================================

def plot_training_curves(
    model_name,
    result,
):

    history = result["history"]

    epochs = [
        x["epoch"]
        for x in history
    ]

    train_accuracy = [
        x["train_accuracy"]
        for x in history
    ]

    val_accuracy = [
        x["val_accuracy"]
        for x in history
    ]

    train_loss = [
        x["train_loss"]
        for x in history
    ]

    val_loss = [
        x["val_loss"]
        for x in history
    ]

    # --------------------------------------------------------
    # Accuracy
    # --------------------------------------------------------

    plt.figure(
        figsize=(9, 6)
    )

    plt.plot(
        epochs,
        train_accuracy,
        marker="o",
        label="Training Accuracy"
    )

    plt.plot(
        epochs,
        val_accuracy,
        marker="o",
        label="Validation Accuracy"
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Accuracy"
    )

    plt.title(
        f"{model_name.upper()} - Accuracy"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    path = os.path.join(
        RESULTS_DIR,
        f"{model_name}_accuracy_curve.png"
    )

    plt.savefig(
        path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()

    # --------------------------------------------------------
    # Loss
    # --------------------------------------------------------

    plt.figure(
        figsize=(9, 6)
    )

    plt.plot(
        epochs,
        train_loss,
        marker="o",
        label="Training Loss"
    )

    plt.plot(
        epochs,
        val_loss,
        marker="o",
        label="Validation Loss"
    )

    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        "Loss"
    )

    plt.title(
        f"{model_name.upper()} - Loss"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    path = os.path.join(
        RESULTS_DIR,
        f"{model_name}_loss_curve.png"
    )

    plt.savefig(
        path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# CONFUSION MATRIX
# ============================================================

def plot_confusion_matrix(
    model_name,
    result,
):

    cm = np.asarray(
        result["confusion_matrix"]
    )

    fig, ax = plt.subplots(
        figsize=(7, 6)
    )

    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=CLASS_NAMES
    )

    display.plot(
        ax=ax,
        values_format="d"
    )

    ax.set_title(
        f"{model_name.upper()} - Confusion Matrix"
    )

    plt.tight_layout()

    path = os.path.join(
        RESULTS_DIR,
        f"{model_name}_confusion_matrix.png"
    )

    plt.savefig(
        path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()

    print(
        "Saved:",
        path
    )


# ============================================================
# PER CLASS F1
# ============================================================

def plot_per_class_f1(
    model_name,
    result,
):

    report = result[
        "classification_report"
    ]

    values = []

    for class_name in CLASS_NAMES:

        values.append(
            report[class_name]["f1-score"]
        )

    plt.figure(
        figsize=(9, 6)
    )

    plt.bar(
        CLASS_NAMES,
        values
    )

    plt.xlabel(
        "Modulation Family"
    )

    plt.ylabel(
        "F1 Score"
    )

    plt.title(
        f"{model_name.upper()} - Per-Class F1"
    )

    plt.ylim(
        0,
        1
    )

    for i, value in enumerate(
        values
    ):

        plt.text(
            i,
            value + 0.02,
            f"{value:.3f}",
            ha="center"
        )

    plt.tight_layout()

    path = os.path.join(
        RESULTS_DIR,
        f"{model_name}_per_class_f1.png"
    )

    plt.savefig(
        path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# PRINT SUMMARY
# ============================================================

def print_summary(df):

    print("\n")
    print("=" * 100)
    print("MODEL PERFORMANCE SUMMARY")
    print("=" * 100)

    display_df = df.copy()

    for column in [
        "Best Validation Accuracy",
        "Test Accuracy",
        "Precision",
        "Recall",
        "F1",
    ]:

        display_df[column] = (
            display_df[column]
            .map(
                lambda x: f"{x:.4f}"
            )
        )

    print(
        display_df.to_string(
            index=False
        )
    )

    best_model = df.loc[
        df["Test Accuracy"].idxmax()
    ]

    print("\n")
    print(
        "Best model:",
        best_model["Model"]
    )

    print(
        f"Accuracy: "
        f"{best_model['Test Accuracy']:.4f}"
    )


# ============================================================
# MAIN
# ============================================================

def main():

    print("\n")
    print("=" * 80)
    print(
        "RADIO ML 2018.01A VISUALIZATION"
    )
    print("=" * 80)

    if not os.path.exists(
        RESULTS_DIR
    ):

        print(
            "\nResults directory does not exist:"
        )

        print(
            RESULTS_DIR
        )

        print(
            "\nRun experiment.py first."
        )

        return

    df = load_comparison()

    if df is None:

        return

    print_summary(
        df
    )

    # ========================================================
    # COMPARISON PLOTS
    # ========================================================

    plot_accuracy_comparison(
        df
    )

    plot_f1_comparison(
        df
    )

    # ========================================================
    # INDIVIDUAL MODEL PLOTS
    # ========================================================

    for model_name in df["Model"]:

        print("\n")
        print(
            "=" * 80
        )

        print(
            f"VISUALIZING: "
            f"{model_name.upper()}"
        )

        print(
            "=" * 80
        )

        result = load_result(
            model_name
        )

        if result is None:

            continue

        plot_training_curves(
            model_name,
            result
        )

        plot_confusion_matrix(
            model_name,
            result
        )

        plot_per_class_f1(
            model_name,
            result
        )

    print("\n")
    print("=" * 80)
    print(
        "VISUALIZATION COMPLETE"
    )
    print("=" * 80)

    print(
        "\nAll figures are saved in:"
    )

    print(
        RESULTS_DIR
    )


if __name__ == "__main__":
    main()
```
